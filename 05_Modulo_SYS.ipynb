```python
import os, sys

lista = list()
for numero in range(1, 100):
  lista.append(numero)

consumol = sys.getsizeof(lista)
print(f"La lista tiene {len(lista)} de elemento y tiene un consumo de {consumol} bytes")

tupla = tuple(lista)
consumot = sys.getsizeof(tupla)
print(f"La tupla tiene {len(tupla)} de elemento y tiene un consumo de {consumot} bytes")

proporcion = (consumot / consumol) * 100
print(f"La tupla es {proporcion} veces mas pequeña que la lista")
```
