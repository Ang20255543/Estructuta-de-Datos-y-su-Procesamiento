```python
import functools

lista_numeros = [1,2,3,4,5,6]

def sumar_dos(valor1, valor2):
    return valor1 + valor2

print("El agregado de la lista mediante 'reduce()' es:")
print(functools.reduce(sumar_dos, lista_numeros))
```

```python
import itertools

lista_numeros = [1,2,3,4,5,6]

def sumar_dos(valor1, valor2):
    return valor1 + valor2

resultado_en_lista = list(itertools.accumulate(lista_numeros,sumar_dos))

print("El agregado de la lista mediante 'accumulate()' es:")
print(resultado_en_lista)
```
