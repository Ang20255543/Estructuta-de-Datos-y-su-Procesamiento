```python
import csv

datos_grabar = dict()

mi_auto1 = ("BMW", 2005)
mi_auto2 =("TOYOTA", 2015)

datos_grabar[10] = mi_auto1
datos_grabar[20] = mi_auto2

print(datos_grabar)

archivo = open("autos.csv", "w", encoding ='latin1', newline="")
grabador = csv.writer(archivo)

grabador.writerow(("Clave", "Marca", "Año"))

for clave, datos in datos_grabar.items():
  grabador.writerow((clave, datos[0], datos[1]))

archivo.close()
```

```python
import csv, sys

data= dict()

try:
  with open("autos.csv", "r", encoding ='latin1', newline="") as archivo:
    lector = csv.reader(archivo)
    next(lector)

    for clave, marca, año in lector:
      data[int(clave)] = (marca, int(año))

except FileNotFoundError:
  print("No se encontró el archivo")

except Exception:
  exception = sys.exc_info()[0]
  print("Ocurrió un error:", exception[0])
  print("Mensaje del error: ", exception[1])

else:
  print(data)
```

```python
import sys

datos_leer = dict()

try:
  with open("autos.csv", "r", encoding ='latin1', newline="") as archivo:
    lector = csv.reader(archivo)
    next(lector)

    for clave, marca, año in lector:
      datos_leer[int(clave)] = (marca, int(año))

except FileNotFoundError:
  print("No se encontró el archivo")

except csv.Error as fallo_csv:
  print("Fallo al leer el archivo CSV:", fallo_csv)

except Exception:
  exception = sys.exc_info()[0]
  print("Ocurrió un error:", exception[0])
  print("Mensaje del error: ", exception[1])

else:
  print(datos_leer)
```
