```python
from datetime import datetime, date, timedelta
import time

hr = datetime.now()
print(f"Tipo de objetos de la hora {type(hr)}")
print(f"La hora es: {hr}")
print(f"la hora de la {hr} es {hr.hour}")
print(f"los minutos de {hr} es {hr.minute}")
print(f"Los segundos de {hr} es {hr.second}")
print(f"Los microsegundos de {hr} es {hr.microsecond}")
```

```python
factual = datetime.today()
print(f"El objeto de la fecha actual es {type(factual)}")
print(f"La fecha actual es {factual}")
print(f"El año actual es {factual.year}")
print(f"El mes actual es {factual.month}")
print(f"El dia actual es {factual.day}")
```

```python
fcapturada = input("Ingrese una fecha en formato dd/mm/aaaa: \n")
fprocesada = datetime.strptime(fcapturada, "%d/%m/%Y").date()
print(f"El objeto de la fecha capturada es {type(fcapturada)}")
print(f"El objeto de la fecha procesada es {type(fprocesada)}")
print(f"La fecha capturada se transformo en {fprocesada}")
```

```python
print(f"De entrada, tenemos el valor {fprocesada} en la variable con el tipo de dato especializado\n")

print(f"El mismo valor lo podemos representar en el formato dd/mm/aaaa: {fprocesada.strftime('%d/%m/%Y')}")
print(f"o tambien en formato mm/dd/aaaa: {fprocesada.strftime('%m/%d/%Y')}")
```

```python
cant_dias= int(input(f"Dime la cantidad de dias a adelantar respacto a {fprocesada}"))
nfecha = fprocesada + timedelta(days=+cant_dias)

print(f"La nueva fecha es: {nfecha}")
```

```python
import os, sys

lista = list()
for numero in range(1, 100):
  lista.append(numero)

consumol = sys.getsizeof(lista)
print(f"La lista tiene {len(lista)} de elemento y tiene un consumo de {consumol} bytes")

tupla = tuple(lista)
consumot = sys.getsizeof(tupla)
print(f"La tupla tiene {len(tupla)} de elemento y tiene un consumo de {consumot} bytes")

proporcion = (consumot / consumol) * 100
print(f"La tupla es {proporcion} veces mas pequeña que la lista")
```
