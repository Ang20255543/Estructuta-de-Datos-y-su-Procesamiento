```python
class cola:
  def __init__(self):
    self.items = []

  def estaVacia(self):
    return self.items == []

  def agregar(self, item):
    self.items.insert(0, item)

  def avanzar(self):
    return self.items.pop()

  def tamaño(self):
    return len(self.items)


c=cola()

c.agregar(4)
c.agregar("perro")
c.agregar(True)
c.tamaño()
```

```python
cola =[]
separar = ("-" * 20) + "\n"

def main():
  print("""1. Insertar cola
2. Borar cola
3. Listar cola
4. Salir""")
  print(separar * 1)

  option = input("Ingrese una opcion: ")

  if str(option) == "1":
    elemento = input("Ingrese un elemento: ")
    cola.append(str(elemento))
    print(separar * 1)
    main()

  elif option == "2":
    if len(cola) > 0:
      cola.pop(0)
      print(separar *1)
      main()
    else:
      print("La cola esta vacia")
      print(separar*1)
      main()

  elif option == "3":
    print("La cola actual es: ")
    print(cola)
    print(separar * 1)
    main()

  else:
    print("Programa terminado")
    print(separar * 1)

main()
```

```python
cola  = list()
for cantidad in range(5):
  nuevo = input("Nombre del recien llegado: ")
  cola.append(nuevo)

print(f"Se agregaron {len(cola)} elementos:")
for elementos in cola:
  print(elementos)

print(f"Actualmente hay {len(cola)} encolados.")


print("Procedemos a quitarlos de la cola")

while cola:
  print(cola.pop(0))


print(f"Actualmente hay {len(cola)}")
```

```python
import collections

cola= collections.deque()

for cantidad in range(5):
  nuevo = input("Nombre del recien llegado: ")
  cola.append(nuevo)

print(f"Se agregaron {len(cola)} elementos:")
for elementos in cola:
  print(elementos)

print(f"Actualmente hay {len(cola)} elementos.")


print("Procedemos a quitarlos de la cola")

while cola:
  print(cola.popleft())

print(f"Actualmente hay {len(cola)} elementos")
```
