```python
import random

print(f"Obteniendo un numero entero aleatorio que puede ir del 0 al 19: {random.randrange(0,20)}")
print(f"Obteniendo un nnumero entero aleatorio que puede ir del 2 al 20 {random.randrange(2, 21, 2)}")
print(f"Numero aleatorio de entre el 1 al 20 {random.randint(1,20)}")
print(f"Numero aleatorio entre 0.0 y 1.0: {random.random()}")
```

```python
ldp = [10,20,30,40,15,25,35,45]
print(f"la lista de prueba es: {ldp}")


print(f"El valor seleccionado aleatoriamente de la lista es: {random.choice(ldp)}")

random.shuffle(ldp)
print(f"La lista ya perturbada/barajada es: {ldp}")
```
