```python
class stack:
  def __init__(self):
    self.items = []
  def is_empty(self):
    return self.items == []
  def push(self, item):
    self.items.insert(0, item)
  def pop(self):
    return self.items.pop(0)
  def print_stack(self):
    print(self.items)

pila = stack()
pila.push("a")
pila.push("b")
pila.push("c")
pila.push("d")
pila.push("e")
pila.print_stack()
pila.pop()
pila.print_stack()
pila.pop()
pila.print_stack()
```

```python
class Pila:
  def __init__(self):
    self.items = []

  def estaVacia(self):
    return self.items == []

  def extraer(self):
    return self.items.pop()

  def incluir(self, item):
    self.items.append(item)

s = Pila()

s.incluir('hola')
s.incluir('verdadero')
s.incluir(1)
s.incluir('falso')
print(s.extraer())
```

```python
stack = []
SEPARA = ("-" * 20) + "\n"

def main():
  print("""1. Apilar
2. Desapilar elemento
3. Mostrar pila
4. Salir """)
  print(SEPARA * 1)
  option = input("Elige una opcion: ")

  if str(option) == "1":
    elemento = input("Introduzca el numero a apilar: ")
    stack.append(elemento)
    print(SEPARA * 1)
    main()

  elif str(option) == "2":
    if len(stack) == 0:
      print("No hay elementos por desapilar")
      main()
    else:
      print(f"**- Elemento {stack.pop()} desapilado.")

  elif str(option) == "3":
    for i in reversed(range(len(stack))):
      print("Pila", stack[i])
    print(SEPARA * 1)
    main()

  elif str(option) == "4":
    print("Gracias por usar el programa")
    exit()

  else:
    print("Opcion no valida")
    main()

main()
```

```python
pila_con_lista = list()

for i in range(5):
  pila_con_lista.append(input("Nombre a agregar: "))
print(f"Se ha agregado {len(pila_con_lista)} elementos a la pila implementada con una lista\n")

while pila_con_lista:
  print(f"Elemento extraido: {pila_con_lista.pop()}")
print(f"Actualmente quedan {len(pila_con_lista)} elementos en la pila implementada con una lista\n")
```
