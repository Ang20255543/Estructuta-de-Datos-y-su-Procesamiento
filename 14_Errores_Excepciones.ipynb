```python
import sys

lista_v = ['x', 0, 4]

for valor in lista_v:
  try:
    print(f"El valor en turno es: {valor}")
    reciproco = 1/ int(valor)
    print(f"El valor reciproco es: {reciproco}")
  except:
    print("Ha ocurrido un error", sys.exc_info()[0], "ocurrio")
```

```python
import sys

lista_v = ['x', 0, 4]

for valor in lista_v:
  try:
    print(f"El valor en turno es: {valor}")
    reciproco = 1/ int(valor)
    print(f"El valor reciproco es: {reciproco}")
  except:
    print("Ha ocurrido un error", sys.exc_info()[0], "ocurrio")
    excepcion = sys.exc_info()
    for elemenetos in excepcion:
      print(elemenetos)
  else:
    print("Todo ha ido bien")
  finally:
    print("Esta linea siempre se ejecutara")
```

```python
import sys

lista_v = ['x', 0, 4]

for valor in lista_v:
  try:
    print(f"El valor en turno es: {valor}")
    reciproco = 1/ int(valor)
    print(f"El valor reciproco es: {reciproco}")

  except ValueError:
    print("Ha ocurrido un error", valor, "valor no valido")
  except ZeroDivisionError:
    print("Ha ocurrido un error", valor, "valor no valido")
  except:
    print("Ha ocurrido un error", sys.exc_info()[0], "ocurrio")
    excepcion = sys.exc_info()
    for elemenetos in excepcion:
      print(elemenetos)
  else:
    print("Todo ha ido bien")
  finally:
    print("Esta linea siempre se ejecutara")
```

```python
n = int(input("Introduce un numero entero entre 1 y 10: "))
file_name = 'tabla-' + str(n) + '.txt'

f = open(file_name, 'w')
for i in range (1,11):
  f.write(str(n) + ' x ' + str(i) + ' = ' + str(n*i) + '\n')
f.close()

```
