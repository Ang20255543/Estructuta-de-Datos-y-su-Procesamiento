```python
import math

vF = float(input("introduce un numero con decimal: "))

print(f"El entero anterior al numero es: {math.floor(vF)}")
print(f"El entero siguiente al numero es: {math.ceil(vF)}")
print (f"La parte trunca del numero es: {math.trunc(vF)}")
```

```python
vN = int(input("introduce un numero entero: "))

print(f"La raiz cuadrada de {vN} es igual a {math.sqrt(vN)}")
print(f"El factorial de {vN} es igual a {math.factorial(vN)}")
```

```python
pw = int(input("Ingrese un numero entero: "))

print(f"El resultado de elevar {vN} a la potencia de {pw} es igual a {math.pow(vN, pw)}")
print(f"El valor de pi es {math.pi}")
```
